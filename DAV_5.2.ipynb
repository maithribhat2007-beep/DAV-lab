```python
import pandas as pd
import numpy as np
```

```python
from scipy.stats import zscore
```

```python
data = {
    "Students" : ["A", "B", "C", "D"],
    "Marks" : [50, 60, 70, 200]
}
df=pd.DataFrame(data)
```

```python
pip install scipy
```

```python
df["Z_Score"] = zscore(df["Marks"])
print(df)
```

```python
outliers=df[abs(df["Z_Score"])>3]
print(outliers)
```

```python
mean=df["Marks"].mean()
std=df["Marks"].std(ddof=0)
lower_limit=mean - (3*std)
upper_limit=mean + (3*std)
```

```python
print(mean)
print(std)
print(lower_limit)
print(upper_limit)
```

```python
df["Marks_Treated"]=df["Marks"].clip(
    lower=lower_limit,
    upper=upper_limit
)
print(df)
```
