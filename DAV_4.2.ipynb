```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Student" : ["A", "B", "C" ,"D", "E"],
    "Marks" : [50, 55, 44, 75, 150]
}
```

```python
df=pd.DataFrame(data)
```

```python
print("original ds")
print(df)
```

```python
Q1=df["Marks"].quantile(0.25)
print(Q1)
```

```python
Q3=df["Marks"].quantile(0.75)
print(Q3)
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_limit = Q1-1.5*IQR
upper_limit = Q1+1.5*IQR
print(lower_limit)
print(upper_limit)
```

```python
outliers = df[
    (df["Marks"]<lower_limit)|
    (df["Marks"]>lower_limit)]
```

```python
print(outliers)
```

```python
df["Marks_Treated" == np.where(
    df["Marks"]>upper_limit,
    upper_limit,
    np.where(
        df["Marks"]<upper_limit,
    lower_limit,
        df["Marks"]))]
```

```python
df_capped =df.copy()
df_capped["Marks"] =df_capped["Marks"].clip(
    lower=lower_limit,
    upper=upper_limit
)
print("after capping")
print(df_capped)
```
